```python
def build_prompt(question, context):
    return f"""Voici des extraits de documents pertinents  :
{context}

Question : {question}
Réponse :"""



def ask_llm(prompt):
    response = client.chat.completions.create(
        model="llama3",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.2,
    )
    return response.choices[0].message.content
    
```

```python
from openai import OpenAI
client = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")

```

```python
question = "Quel est un plan typique de la caro-khan?"
context = "fused_fused_fused_document_0_1_2: Chaque ligne donne lieu à des plans stratégiques spécifiques. | Les joueurs qui choisissent la Caro-Kann doivent bien connaître ces subtilités pour ne pas se retrouver passifs. | Stratégie et plans typiques | Un plan courant consiste à jouer Bf5, e6, Nd7, Ngf6, et c5 ou c6 selon la structure."
prompt = build_prompt(question,context)
print(ask_llm(prompt))
```
